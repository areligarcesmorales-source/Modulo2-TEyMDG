## **EJEMPLO.**
Sea el experimento de lanzar una moneda 3 veces, de manera independiente, es decir, el $i$-esimo lanzamiento no afecta al $i+1$ lanzamiento. Entonces el espacio muestral asociado a este experimento es:

$$ \Omega = \{A,S\} \times \{A,S\} \times \{A,S\} = \{A,S\}^3 $$

$A=\{\text{aguilas}\}, A=\{\text{soles}\} $ y sabemos que $|Omega|= 2^3 =8$

Sean los eventos:

- $A$ : todos los resultados son aguila
- $S$ : todos los resultados son sol
- $E$ : por lo menos un resultado es sol (1 o 2 o 3 soles)

Calcular las probabilidades por medio de la probabilidad Laplaciana

Para hacer esto, vamos a descomponer el evento $A$ de la siguiente manera

$$ A = A_1  \cap A_2 \cap A_3 $$

donde $A_i$ es que caiga águila en el $i$esimo lanzamiento

## CREANDO EL ESPACIO MUESTRAL CON  ITERTOOLS

```python
#Importamoe el modulo 
import itertools
from itertools import product
```

```python
Omega = set(product({'A','S'}, repeat =3))  ## Set convierte en conjutno y product es el producto cartesiano
Omega
```

```python
##La cardinalidad de Omega es 
len(Omega)
```

Vamos a crear el evento $A$

```python
## Para A_1 :el primer elemento es aguila
A_1 = { om for om in Omega if om[0]=='A'}
A_1
```

```python
## Para A_2 :el segundo elemento es aguila
A_2 = { om for om in Omega if om[1]=='A'}
A_2
```

```python
## Para A_3 :el stercer elemento es aguila
A_3 = { om for om in Omega if om[2]=='A'}
A_3
```

```python
# Voy a interceptar A_1, A_2 y A_3
A =A_1.intersection(A_2.intersection(A_3))
A
```

```python
 card_A = len(A)
```

```python
P_A = card_A / len(Omega)
P_A
```

Si queremos el resultado en friccion, utilizar fractions

```python
from fractions import Fraction
```

```python
## Vamos a crear la funcion de probabilidad
def p(E,Omega):
    p = Fraction(len(E),len(Omega))
    return p
```

```python
Pro_A = p(A,Omega)
Pro_A
```

## PROBABILIDAD CONDICIONAL

Si añadimos información condiciones a nuestro experimento aleatorio, entonces la probabilidad de un evento cambia

Definamos la **probabilidad condicional** de que ocurra un evento $A$ sabiendo que ocurre un evento $B$ de probabilidad positiva, entonces

$$ \mathbb{P}(A \mid B ) = \frac{\mathbb{P}(A\cap B)}{\mathbb{P}(B)} = \frac{|A\cap B|}{|B|} $$

```python
## Vamos a definir la funcion de probabilidad condicional

def prob_cond (E, F):
    return len(E & F) / len(F)
```

Quereos calcular la probabilidad condicional de los siguiente eventos

- $B$: el primer lanzamiento es aguila
- $C$: dos de los resultados son aguila (exactamente)

Quiero calcular $\mathbb{P}(B\mid C)$

```python
# Para el evento B : el primer lanzamiento es águila

B = { om for om in Omega if om[0]=='A'}
B
```

```python
## Para el evento C : dos de ñps resultados son aguila

C = {om for om in Omega if om.count('A')==2}
C
```

```python
# Calculamos la P(B|C)

prob_cond(B,C)
```

```python
prob_cond(C,B)
```

## Independencia de eventos

Decimos que los eventos $A$ y $B$ son independientes si 

$$\mathprob{P}(A/cap B)

```python
## Definimos la funcion de probabilidad de un evento
def prob(E):
    return len(E) / len(Omega)
```

```python
## Definimos la funcion de probabilidad de un evento

def indep(E,F):
    return prob(E & F)== prob(E) * prob(F)
```

```python
## Son B y C eventos indepentientes?
indep(B,C)
```
